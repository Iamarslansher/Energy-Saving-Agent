```python
!pip install -U google-genai
from google import genai

from google.colab import userdata
api_key = userdata.get("agent-key")

client = genai.Client(api_key=api_key)
MODEL = "models/gemini-3.5-flash"

students = [

    {
        "name": "Arsalan Sher",
        "program": "BSCS",
        "semester": 5,
        "section": "C",
        "subject": "SPM",
        "marks": 85
    },

    {
        "name": "Ali Ahmed",
        "program": "BSCS",
        "semester": 5,
        "section": "C",
        "subject": "SPM",
        "marks": 78
    },

    {
        "name": "Usman Khan",
        "program": "BSCS",
        "semester": 5,
        "section": "B",
        "subject": "AI",
        "marks": 91
    },

    {
        "name": "Hamza Malik",
        "program": "BSCS",
        "semester": 4,
        "section": "A",
        "subject": "DBMS",
        "marks": 74
    },

    {
        "name": "Ahmed Raza",
        "program": "BSCS",
        "semester": 6,
        "section": "C",
        "subject": "Web Engineering",
        "marks": 88
    },

    {
        "name": "Bilal Hussain",
        "program": "BSSE",
        "semester": 5,
        "section": "A",
        "subject": "Software Engineering",
        "marks": 82
    },

    {
        "name": "Zain Ali",
        "program": "BSCS",
        "semester": 5,
        "section": "B",
        "subject": "Data Mining",
        "marks": 76
    },

    {
        "name": "Hassan Khan",
        "program": "BSIT",
        "semester": 3,
        "section": "A",
        "subject": "Programming Fundamentals",
        "marks": 89
    },

    {
        "name": "Saad Ahmed",
        "program": "BSCS",
        "semester": 6,
        "section": "B",
        "subject": "AI",
        "marks": 93
    },

    {
        "name": "Muhammad Umar",
        "program": "BSCS",
        "semester": 4,
        "section": "C",
        "subject": "Computer Networks",
        "marks": 81
    }

]

def  search_marks(name, semester, section, subject):
  for student in students:
    if(
            student["name"].lower() == name.lower() and
            student["semester"] == semester and
            student["section"].lower() == section.lower() and
            student["subject"].lower() == subject.lower()
            ):
      return student
  return None

def gemini_result(name, semester, section, subject):
    prompt = f"""
You are a FUUAST Marks Assistant.

Important rules:

 1. You do not have access to the official FUUAST database.
 2. Never invent or guess student marks.
 3. If a student record is not available in the local database,
    clearly tell the user that the record was not found.
4. Give a short and student-friendly response.

User request:
{name, semester, section, subject}
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text

print("🎓 FUUAST Marks Checker Agent Started")
print("Type 'exit' to stop.\n")

while True:
  name = input("Name: \n")
  semester = int(input("Semester: \n"))
  section = input("Section: \n")
  subject = input("Subject: \n")
  # if user_input.lower() == "exist"
  #  break

  result = search_marks(name, semester, section, subject)
  if result:
     print(
            f"\nAgent: {result['name']} "
            f"got {result['marks']} marks in "
            f"{result['subject']}.\n"
        )
  else:
     reply = gemini_result(name, semester, section, subject)

     print("\nAgent:", reply, "\n")




print("Executed!")
```
