```python

flights = [
    {
        "flight_name": "Blue Airline",
        "destination": "Lahore",
        "departure": "Karachi",
        "seats": "full"
    },
    {
        "flight_name": "Pakistan Airline",
        "destination": "Islamabad",
        "departure": "Karachi",
        "seats": "full"
    },
    {
        "flight_name": "Redline Airline",
        "destination": "Karachi",
        "departure": "Multan",
        "seats": "available"
    },
    {
        "flight_name": "Airline Express",
        "destination": "Lahore",
        "departure": "Islamabad",
        "seats": "available"
    },
    {
        "flight_name": "KPK Airline",
        "destination": "KPK",
        "departure": "Karachi",
        "seats": "full"
    }
]

print("✈️ Welcome to Jinnah Airport!")
print("How can we assist you?\n")

destination = input("Please enter your destination: ").strip().lower()
departure = input("Please enter your departure city: ").strip().lower()

available_flights = []
for flight in flights:
    if (
        flight["destination"].lower() == destination
        and flight["departure"].lower() == departure
        and flight["seats"].lower() == "available"
    ):
        available_flights.append(flight)

if available_flights:

    print("\n✅ Flight Found!")
    for flight in available_flights:
        print(
            f"✈️ {flight['flight_name']} "
            f"| {flight['departure']} → {flight['destination']} "
            f"| Seats: {flight['seats']}"
        )

else:

    print("\n❌ Sorry!")
    print(
        f"No flight with available seats was found "
        f"from {departure.title()} to {destination.title()}."
    )
```
